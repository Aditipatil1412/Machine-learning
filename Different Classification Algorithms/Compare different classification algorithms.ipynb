## Classification Algorithms Comparison

Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

Load Iris Dataset

```python
iris = load_iris()

X = iris.data
y = iris.target

print("Dataset loaded successfully!")
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Classes:", iris.target_names)
```

Create DataFrame

```python
df = pd.DataFrame(
    X,
    columns=iris.feature_names
)

df["Species"] = y

df.head()
```

Dataset Information

```python
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
df.info()
```

Check Missing Values

```python
print("Missing Values:")
print(df.isnull().sum())
```

Check Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

Display Statistical Summary

```python
df.describe()
```

Data Visualization

```python
## Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Iris Dataset")
plt.show()
```

Train-Test Split

```python
## Split the Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

Feature Scaling

```python
## Standardize Features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

Define Classification Models

```python
## Create Models
models = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),
    
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),
    
    "SVM": SVC(
        kernel="rbf",
        random_state=42
    ),
    
    "KNN": KNeighborsClassifier(),
    
    "Naive Bayes": GaussianNB()
}
```

Train and Evaluate Models

```python
##Train All Models
results = []

predictions = {}

for name, model in models.items():
    
    model.fit(X_train_scaled, y_train)
    
    y_pred = model.predict(X_test_scaled)
    
    predictions[name] = y_pred
    
    accuracy = accuracy_score(y_test, y_pred)
    
    precision = precision_score(
        y_test,
        y_pred,
        average="macro"
    )
    
    recall = recall_score(
        y_test,
        y_pred,
        average="macro"
    )
    
    results.append([
        name,
        precision,
        recall,
        accuracy
    ])
```

Comparison Table

```python
## Display results
results_df = pd.DataFrame(
    results,
    columns=[
        "Algorithm",
        "Precision",
        "Recall",
        "Accuracy"
    ]
)

results_df
```

```python
##Round the Results
results_df[
    ["Precision", "Recall", "Accuracy"]
] = results_df[
    ["Precision", "Recall", "Accuracy"]
].round(3)

results_df
```

Accuracy Comparison

```python
plt.figure(figsize=(6,3))

plt.bar(
    results_df["Algorithm"],
    results_df["Accuracy"]
)

plt.xlabel("Classification Algorithm")
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison of Classification Algorithms")

plt.xticks(rotation=30)

plt.show()
```

Precision Comparison

```python
plt.figure(figsize=(6,3))

plt.bar(
    results_df["Algorithm"],
    results_df["Precision"]
)

plt.xlabel("Classification Algorithm")
plt.ylabel("Precision")
plt.title("Precision Comparison of Classification Algorithms")

plt.xticks(rotation=30)

plt.show()
```

Recall Comparison

```python
plt.figure(figsize=(6,3))

plt.bar(
    results_df["Algorithm"],
    results_df["Recall"]
)

plt.xlabel("Classification Algorithm")
plt.ylabel("Recall")
plt.title("Recall Comparison of Classification Algorithms")

plt.xticks(rotation=30)

plt.show()
```

Confusion Matrix

```python
## SVM Confusion Matrix
svm_predictions = predictions["SVM"]

cm = confusion_matrix(
    y_test,
    svm_predictions
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
)

disp.plot()

plt.title("Confusion Matrix - SVM")
plt.show()
```

Find the best model

```python
best_accuracy_model = results_df.loc[
    results_df["Accuracy"].idxmax(),
    "Algorithm"
]

best_precision_model = results_df.loc[
    results_df["Precision"].idxmax(),
    "Algorithm"
]

best_recall_model = results_df.loc[
    results_df["Recall"].idxmax(),
    "Algorithm"
]

print("Highest Accuracy :", best_accuracy_model)
print("Highest Precision:", best_precision_model)
print("Highest Recall   :", best_recall_model)
```

Conclusion

The performance of six classification algorithms was compared using Accuracy, Precision, and Recall. Based on the obtained evaluation results, the classifier with the highest overall performance is considered the most suitable for the Iris dataset. The experiment demonstrates the importance of comparing different classification algorithms before selecting a model.
